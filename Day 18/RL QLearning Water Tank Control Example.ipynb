```python
import numpy as np
import random
```

```python
#Step 2: Define States and Actions
states=np.arange(0,101,10)   # Create an array of integers from 0 to 100 with step size 10
                            # This represents the possible states (water levels) in our problem
actions=['FILL','STOP']
```

```python
print(states)
print(actions)
```

```python
#step 2: create QTable set alpha,gama,epsilon,episodes
#Q table is [state,action]
Q=np.zeros((len(states),len(actions)))

alpha=0.1 #Learning rate
gama=0.9  #Discount Factor
epsilon=0.2  #Exploration Rate
episodes=300  #Training runs

print('Q-Table shape: ',Q.shape)
```

```python
# Step 3: write Reward function
def get_reward(level,action):
    if 40<=level <=70:
        reward =10 # ideal range
    #else:
        #reward =-10 # too low/high 
    if action=='FILL'and level>=90:
        reward -= 10 # overflow risk
    if action=='STOP'and level<=10:
        reward -= 10 # Empty risk
    return reward

print('Reward Example (level=60, FILL): ',get_reward(60,'FILL'))
```

```python
#step 4: Environment Dynamic Change (eg. Water Level Change)

def next_level (level,action):
    if action =='FILL':
        level +=random.choice([5,10,15])
    else:
        level -=random.choice([5,10,15]) 
    return int(np.clip(level,0,100))

print('Next Level Example : ',next_level(50,'FILL'))
```

```python
#Step 5 Training Loop
for ep in range(episodes):
    level=random.choice(states)
    done=False
    for _ in range(1): #limit steps per episode
        #choose action (epsilon-greedy)
        if random.uniform(0,1)<epsilon:
            action=random.choice(actions)
        else:
            action=actions[np.argmax(Q[level//10])]

        next_state=next_level(level,action)
        reward=get_reward(next_state, action)
        
        a=actions.index(action)
        best_next=np.max(Q[next_state//10])
        Q[level//10,a] += alpha*(reward+gama+best_next-Q[level//10, a])
# Q(state, action) = Q(state, action) + α * [reward + γ * max(Q(next_state, all_actions)) - Q(state, action)]

        temp=next_state
print('Training Done')
```
