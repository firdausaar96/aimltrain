```python
import numpy as np
import random
```

```python
#Step 2: Define States and Actions
states={
    0: 'Empty road',
    1: 'Light traffic',
    2: 'Moderate traffic',
    3: 'Heavy traffic',
    4: 'Very heavy traffic'
}
actions=['GREEN','RED']
```

```python
print("Defined traffic states:")
for key, value in states.items():
    print(f"  {key}: {value}")
print("\nDefined possible actions:", actions, '\n')
```

```python
Q=np.zeros((len(states),len(actions)))

alpha=0.1 #Learning rate
gama=0.9  #Discount Factor
epsilon=0.2  #Exploration Rate
episodes=300  #Training runs

print('Q-Table shape: ',Q.shape)
```

```python
def get_reward(traffic,action):  
    
    if action=='GREEN'and traffic==3:
        return 10 # Good:Clear congestion
    elif action=='RED'and traffic==3:
        return -10 # Bad: creates jams
    elif action=='RED'and traffic==0:
        return 5 # Saves Energy
    elif action=='GREEN'and traffic==0:
        return -5 # Waste Power
    elif  traffic==2:
        return 1 # Waste Power
    else:
        return 1
```

```python
#Simulate How traffic change randomly
#traffic can increase, decrease, or stay constant

def next_traffic (traffic,action):
    if action =='GREEN':
        traffic -=random.choice([-1,0,1])
    else:
        traffic +=random.choice([-1,0,1]) 
    return int(np.clip(traffic,0,4))
```
