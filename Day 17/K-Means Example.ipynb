# K-means Clustering Example

```python
#Unsupervised Machine Learning Technique
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
```

```python
x = np.array([
    [7.5, 8.0],
    [9.0, 7.0],
    [10.5, 6.5],
    [6.0, 9.0],
    [13.0, 5.5],
    [8.0, 8.5],
    [14.0, 4.5],
    [9.5, 7.0],
    [7.0, 9.0],
    [12.0, 6.0]
    ])

df=pd.DataFrame(x, columns=['EntryTime','Duration'])
df
```

```python
#create Model
# Create a K-Means clustering model with 3 clusters
# n_clusters=3 specifies we want to group data into 3 clusters
# random_state=0 ensures reproducibility of results
model=KMeans(n_clusters=3, random_state=0)
```

```python
# Apply KMeans clustering to the dataset using 'EntryTime' and 'Duration' features
# The cluster assignments are stored in a new column called 'Cluster'
df['Cluster'] = model.fit_predict(df[['EntryTime', 'Duration']])

# Extract the cluster centers (centroids) from the fitted model
centers = model.cluster_centers_
df
```

```python
plt.figure(figsize=(7,5))
plt.scatter(df['EntryTime'], df['Duration'], c=df['Cluster'])
plt.scatter(centers[:,0], centers[:,1], marker='X', s=200)
plt.xlabel('Entry Time (Hour of Day)')
plt.ylabel('Duration in Office (Hours)')
plt.title('Working Hour Patterns using K-Means')

plt.grid(True)
plt.show()
 
```

```python
plt.figure(figsize=(7,5))
plt.scatter(df['EntryTime' ], df['Duration'], c=df['Cluster'], s=100)
# Plot the centroids of each cluster as red X markers with size 200 and label them as 'Centroid'
# Extract x-coordinates (entry times) and y-coordinates (durations) from the cluster centers array
plt.scatter(centers[:,0], centers[:,1], marker='X', s=200,c='red',label='Centroid')
plt.xlabel('Entry Time (Hour of Day)')
plt.ylabel('Duration in Office (Hours)')
plt.title('Working Hour Patterns using K-Means')
plt.legend()
plt.grid(True)
plt.show()
```

# Lab: Customer Segmentation using K-Means
**Goal** : Group customers by Annual Income and Spending Score.
 
1. Download customer.csv from https://github.com/salman1256/aimltraining/tree/main/Day-17(II)
2. Load data from customers.csv
3. Visualize the data
4. Cluster
5. Interpret each segment like 'High income / high spender' etc.
6. Create Scatter plot for the same


```python
df=pd.read_csv('customers.csv')
df.head()
```

```python
Model=KMeans(n_clusters=5, random_state=0)
```

```python
df['Cluster'] = Model.fit_predict(df[['AnnualIncome', 'SpendingScore']])

# Extract the cluster centers (centroids) from the fitted model
centers = Model.cluster_centers_
df
new_cust=pd.DataFrame({'AnnualIncome':[14], 'SpendingScore':[83]})
new_label=Model.predict(new_cust)[0]
print(new_label)
```

```python
plt.figure(figsize=(7,5))
plt.scatter(df['AnnualIncome' ], df['SpendingScore'], c=df['Cluster'], s=100)
# Plot the centroids of each cluster as red X markers with size 200 and label them as 'Centroid'
# Extract x-coordinates (entry times) and y-coordinates (durations) from the cluster centers array
plt.scatter(centers[:,0], centers[:,1], marker='X', s=200,c='red',label='Centroid')
plt.xlabel('Annual Income')
plt.ylabel('Spending Score')
plt.title('Spending Patterns using K-Means')
plt.legend()
plt.grid(True)
plt.show()
```
