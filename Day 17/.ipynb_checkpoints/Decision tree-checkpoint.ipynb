```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier, plot_tree
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data_promo={
    'Experience_Years':[2,4,5,3,6,5],
    'Performance_Rating':[3,4,5,3,4,4],
    'Promoted':['No','Yes','Yes','No','Yes',None]
}

df=pd.DataFrame(data_promo)
display(df)
```

```python
# Cleaning data
train_df=df.dropna()
train_df
```

```python
#Split data and find input and output
x=train_df[['Experience_Years','Performance_Rating']]
y=train_df['Promoted']
```

```python
#create the model
model=DecisionTreeClassifier()
```

```python
#Train the model
model.fit(x,y)
```

```python
pred_promote=model.predict([[5,4]])[0]
print('Promotion prediction: ',pred_promote)
```

```python
plt.figure(figsize=(6,4))
plot_tree(model,
          feature_names=['Experience_Years','Performance_Rating'],
          class_names=['No','Yes'],
          filled=True)
plt.title('Decision Tree=Employee Promotion')
plt.show()
```

```python
exp_years=int(input('Enter Staff Experience'))
per_rating=int(input('Enter Rating Performance'))
our_prediction=pd.DataFrame([[exp_years,per_rating]],columns=['Experience_Years','Performance_Rating'])
prediction=model.predict(our_prediction)
print(f'\nExperience years: {exp_years} \nPerformance Rating: {per_rating} \nPrediction: {prediction[0]}')
```
