```python
def get_reward(traffic,action):
     
    if action=='GREEN'and traffic==3:
        reward += 10 # Good:Clear congestion
    if action=='RED'and traffic==3:
        reward -= 10 # Bad: creates jams
    if action=='RED'and traffic==0:
        reward += 5 # Saves Energy
    if action=='GREEN'and traffic==0:
        reward -= 5 # Waste Power
    if (action=='GREEN'or 'RED') and traffic==2:
        reward += 1 # Waste Power
    return reward

print('Reward Example (Heavy Traffic (3), GREEN): ',get_reward(0,'RED'))
```
